3. Förklara följande:

a) Vad är syftet med att dela upp data i träningsdata, valideringsdata och
testdata?

The data is split to prevent overfitting to only the given data, such that we can test how well the model handles unknown data.

b) Vad är k-delad korsvalidering för något?

K-fold Cross validation is a method of separating validation and training data. The method separates the data into 'folds', where in 1 fold is used for validation, and the rest of the folds for training. Which fold is used for training then changes, until all folds have been used for validation and trainging respectively.

c) Vad är root mean squared error (RMSE) för något?

RMSE is a method used to calculate the error predicted by a model. As the name suggests the Error of the result predicted by the model for each value is squared, then the mean is taken for each value calculated, then the root of this mean is calculated, giving a more intuitive error value as opposed to just calculating the MSE.

d) Vad är en hyperparameter för något? Vad är en parameter för något?

A hyperparameter is a provided variable which sets how the model should handle the training data. A parameter on the other hand is a variable which emerges in the ML model in response to the training.

A grid search is a method of finding the optimal set of hyperparameters, which tests a given set of variables against all given hyperparameters using cross validation. Using refit means that the best set of hyperparameters as found by the search will be set.

f) Vad är kategorisk data och hur hanteras det? I ditt svar, använd begreppen
nominal data, ordinal data, one-hot-encoding, dummy-variable-encoding
och ordinal encoding.

Categorical data, as the name implies, is non-numerical data, stored as categories. This data can be either nominal, meaning it that the categories can not be ranked, or it can be treated as ordinal, meaning that there is a specific hierarchy to the categories, for example, the age categories of young, middleaged, and old. Depending on the form of this data, there are different methods of encoding this data for use in an ML. Ordinal encoding being used for ordinal data, whereas one-hot-encoding, and dummy-variable-encoding, are similarly ways to turn nominal data into numerical values.

g) Vad är feature engineering för något?

Feature engineering is the selection and preparing data for processing by the learning algorithm.

h) Vad menas med principle of parsimony?

The principle of Persimony, as relates to machine learning, is a principle to make the model as simple as possible, while this simple model can achieve the needed goals of the model. Meaning fewer variables and parameters where possible.

4. Vad menas med att “en modell är en förenkling av verkligheten”?

It means that a model, regardless of complexity, is always going to give a simplified model of reality, and is never going to be able to perfectly predict the real data and scenarios it is trained to predict to perfect accuracy.

5. Vad menas med att en modell är “överanpassad” eller overfitted på engels-
ka?

For a model to be overfitted means for it to be overly specific to the given data, meaning that while it is able to predict the given data it was trained on, it is unable to predict unknown data, or does so incorrectly, instead providing a biased answer based on its training data.

6. Högre är bättre i scikit-learn scoring, vad innebär det?

It means that one needs to account for it when assessing which models are most accurate, as the results we get for methods such as RMSE give a higher value where the amount of error the model predicts as opposed to the testdata is higher, which is typically the opposite of what we want (lower errors). This can be done by instead calculating the negative RMSE in this case.

7. Vad är tvärsnittsdata, tidsseriedata och paneldata? Exemplifiera när re-
spektive datakategori kan uppstå

Cross sectional data is a cross section of a specific timepoint, such as for houseprices in a specific area, at a given data.
Time series data, follows data collected over a given time period, given the previous example, this would instead be the houseprices as followed over time, with constant data collected at multiple points, such that we can follow the change.
Panel data is data which tracks the same subjects over time, while also having miltiple different forms of data collected. Given the previous example, this would be following the same houses over time, keeping track of their price changes, as well as other forms of data, such as, for example, their water bills.

9. Generellt sett gäller det att högre är bättre i scikit-learn scoring, därför
används exempelvis scoring='neg_mean_squared_error'. Förklara lo-
giken bakom detta, det vill säga att vi använder “negative” mean squared
error.

Like explained a bit earlier, the scikit method ranks the scores as higher being better, but the MSE method mentioned here assignes a value based on the mean (squared) error predicted by the model, as compared to the truth in our training data, which. This behaviour is the opposite of what we want, as we want the models to be decided on how accurate they are, with more accurate being better, meaning a lower MSE score. To work around this we therefore use neg_mean_squared_error, which gives the negative of the result calculated by the model, which makes the scikit method behave as desired.
