# Theory

1. Vad menas med curse of dimensionality.

That as the dimensions of the data increases, the data becomes more unpredictable. In effect this means that it becomes harder to make predictions about it, and that it requires more datapoints to learn general patterns in the data.

2. Vad är dimensionsreducering och varför görs det?

Dimension reduction is to reduce a dataset to a lower dimension, meaning to map a higher number of variables to a lower number of variables.
This can be done for multiple reasons, such as wanting to visualize it or to reduce the time it takes to train the model, and, more importantly, it is also in some cases done to improve the prediction of patterns within the data, as this is effectively also a way to reduce the variance, but increasing the bias, preventing overfitting by throwing away outliers.

3. Förklara översiktligt hur PCA fungerar. Använd figur 5.4 på sidan 224 i din förklaring.

Principal component analysis, as the name implies, is a way of mapping the primary components of the data, reducing a higher dimensional dataset to its principal components. Figure 5.4 shows how the PCA determines these principal components in practice, first determining the axis within the high dimensional dataset which retains the most variance C1 (Typically not aligned with any existing axis corresponding with the dataset). It then selects an axis which is ortogonal to the first line as its next principal component C2. This continues until the principal component for as many dimensions the dataset contains. How many principal components are kept is then determined either as a perentage of how much variance should be maintained, or a fixed number of dimensions. We see the result of the dimensional reduction for each of the cross-axes through the leftmost 2 dimensional data, to a 1 dimensional representation on the right.

4. Hur kan kernel PCA utvärderas?

Kernel PCA is a PCA using the same kernel tricks which allows us to use some other linear techniques to classify polynomial data, to perform a non-linear projection of the data. To evaluate a kernel PCA the same methods we have encountered before, using a grid search to evaluate which kernels are best, providing the grid search function an excecution pipeline including both the kernel pca object, and an evaluation model, such that the model can be evaluated during the gridsearch.

så bra prediktioner som möjligt. Kalle påstår att det inte riktigt stämmer
eftersom tid också är en viktig aspekt. Både för själva modellträningen
och för själva prediktionerna. Vad säger du?

While optimally we would want to always have the best model, in a real world scenario, there are other factors, such as time, which also play an important role in deciding how a model is trained and constructed, say for example if the model being trained is to be deployed for a time sensitive purpose, taking excessive time to train it, would may then result in it being obsolete by the time training is finished. It is exactly this that PCA can be of help in, reducing the dimensionality, and therefore also the time it takes for models to train and evaluate the data. Given that the predictions remain sufficient for its intended purpose, the reduction in accuracy may then be a worthwhile sacrifice.

6. Efter att vi genomfört en PCA, vad händer med tolkningen av variablerna?

Since the principal components are a combination of multiple predictor variables the interpretability of the variables themselves innevitably gets reduced, as they are no longer dirrectly linked to just a single variable.
