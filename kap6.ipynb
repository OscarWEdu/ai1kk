Clustering is a ml model which is used to discover clusters of data within a dataset. This can be used to label and categorize data, which may be helpful, for example, when categorizing different types of customers, allowing a company segregate their customer segments, or to discover outliers, which may be important where standardization is important, or in discovering potential errors. 2. Förklara översiktligt hur K-means fungerar. Använd figur 6.3 (sidan 238)
och figur 6.4 (sidan 239) i din förklaring.

K-means is one of the simpler methods of discovering clusters in the data. When clusters are calculated in a specified number of centroids, the centroids are either randomly placed within the n-dimensional dataset, or choosen from one of the datapoints, depending on the initialization method. The datapoints are then assigned a cluster depending on which centroid is the closest. This is an iterative process, where the centroid is then moved to the mean position of the cluster, and the clusters re-computed based on he new centroid positions.
Typically this is repeated a number of times, with different initial centroid locations, to reduce the likelyhood of producing sub-optimal results due to the centroids being stuck at local mean plateaus.

Because of how the clusters are computed based on a mean of the datapoints within it, the k-means model performs poorly on non-spherical clusters, as we can see when looking at the upper righthand graph of the estimated result in figure 6.4 against the ground truth in figure 6.3. It also struggles with large variance differences between the clusters, and large size difference between the clusters, as can be seen in the lower left corner and the lower right corner respectively.
The upper left graph shows how the clustering produces incorrect estimations when specifying the wrong number for clusters.

3. Hur kan man välja vilket antal kluster som ska användas för en K-means
modell? Använd inertia, silhouette score och silhouette diagram i ditt svar.

There are 3 different ways to estimate the optimal number of clusters for the k-means model (as far as is covered by the material); inertia, silhouette score, and a silhouette diagram.
Inertia calculates the mean distance between the points of the clusters and their centroids, while this can be usefull for displaying large differences due to the number of clusters specified, but means that a higher number of clusters will always produce a lower (better) inertia score.
One way to avoid this drawback, is to calculate the Silhouette coefficient of each point, which is a way of measuring how well a point fits within its cluster by calculating the mean value of the distance of the point to all other points within its cluster, minus the mean distance to all points of the closest cluster. (This is then also divided by the highest of those 2 values to get a value between -1 and 1).
The silhouette score is the mean value of the silhouette coefficient for each point, while the silhouette diagram is a way to visualize the silhouette coefficient of all the points, essentially a bar chart of the values, sorted by cluster. where the silhouette score is added as a simple line to reference the groupings of bars against. As even the silhouette score may not necessarily result in the most representative number of clusters, the silhouette chart can instead be used to detect certain patterns, helping us choose the best number, such as typically not picking cluster numbers where the some clusters do not have values which exceed the silhouette score.
